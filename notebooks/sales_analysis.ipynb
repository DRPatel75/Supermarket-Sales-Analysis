```python
import pandas as pd
import numpy as np

df = pd.read_csv("D:\Projects\Supermarket-Sales-Analysis\data\supermarket_sales.csv")

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Date'].head()
```

```python
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day_name()
```

```python
df[['Date', 'Month', 'Day']].head()
```

```python
df.duplicated().sum()
```

```python
total_revenue = df['Total'].sum()

print("Total Revenue:", total_revenue)
```

```python
total_quantity = df['Quantity'].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
average_sale = df['Total'].mean()

print("Average Sale:", average_sale)
```

```python
total_transactions = df['Invoice ID'].nunique()

print("Total Transactions:", total_transactions)
```

```python
category_sales = (
    df.groupby('Product line')['Total']
      .sum()
      .sort_values(ascending=False)
)

print(category_sales)
```

```python
top_categories = category_sales.head(5)

print(top_categories)
```
